```python
# Install required libraries
!pip install -q nltk spacy
!python -m spacy download en_core_web_sm -q
import nltk
import spacy
import pandas as pd
from nltk.sentiment import SentimentIntensityAnalyzer

# Download VADER
nltk.download("vader_lexicon", quiet=True)

# Load NLP models
nlp = spacy.load("en_core_web_sm")
sia = SentimentIntensityAnalyzer()

# Textual data
texts = [
    "I really enjoyed the new iPhone launched by Apple in California.",
    "The customer service of Microsoft was excellent.",
    "I am disappointed with the poor quality of this product.",
    "Rahul bought a new laptop from Amazon in Delhi."
]

# Store results
results = []
for text in texts:

    # Text preprocessing
    clean_text = text.lower()

    # Sentiment analysis
    score = sia.polarity_scores(clean_text)["compound"]
    if score >= 0.05:
        sentiment = "Positive"
    elif score <= -0.05:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    # Named Entity Recognition
    doc = nlp(text)
    entities = []
    for ent in doc.ents:
        entities.append(f"{ent.text} ({ent.label_})")
    results.append({
        "Text": text,
        "Sentiment": sentiment,
        "Score": round(score, 2),
        "Named Entities": ", ".join(entities) if entities else "None"
    })

# Display results
df = pd.DataFrame(results)
print("NLP TEXT ANALYSIS")
print("=" * 80)
display(df)
```
